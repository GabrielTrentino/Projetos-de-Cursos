## Exercício 1

A empresa 'Componentes Inteligentes Ltda.' atua no setor de manufatura de peças eletrônicas de alta precisão. Tarefa
Desenvolva um programa que calcule o custo de produção total de um lote e o custo por unidade de um componente eletrônico, considerando o preço e a quantidade de dois componentes básicos e um custo fixo.

Entrada
O programa receberá, nesta ordem, seis valores numéricos. O primeiro é um valor de ponto flutuante representando o custo unitário do componente_a. O segundo é um número inteiro indicando a quantidade de componente_a utilizada em cada produto final. O terceiro é um valor de ponto flutuante para o custo unitário do componente_b. O quarto é um número inteiro para a quantidade de componente_b utilizada por produto. O quinto é um valor de ponto flutuante que representa um custo fixo total para todo o lote de produção. Por fim, o sexto valor é um número inteiro que corresponde à quantidade total de produtos fabricados neste lote. Todos os valores serão positivos.

Saída
O programa deve exibir dois valores. O primeiro valor é o custo total do lote de produção, apresentado como um número de ponto flutuante. O segundo valor é o custo por unidade do produto final, também como um número de ponto flutuante, formatado para exibir exatamente duas casas decimais. Os valores devem ser apresentados um abaixo do outro.

```
c_unitario_componente_a = float(input('Digite o custo unitário do componente A: '))
quantidade_componente_a = int(input("Digite a quantidade do compomente A utilizada em cada produto final: "))
c_unitario_componente_b = float(input('Digite o custo unitário do compomente B: '))
quantidade_componente_b = int(input("Digite a quantidade do compomente B utilizada em cada produto final: "))
custo_fixo_lote = float(input("Digite o custo fixo total para todo o lote: "))
quantidade_total = int(input("Digite a quantidade total de produtos fabricados neste lote: "))

custo_a = c_unitario_componente_a * quantidade_componente_a
custo_b = c_unitario_componente_b * quantidade_componente_b

custo_lote = (custo_a + custo_b) * quantidade_total + custo_fixo_lote
custo_unitario = custo_lote/quantidade_total

print(f"Custo total do lote de produção: R${custo_lote}")
print(f"Custo por unidade do produto final: R${custo_unitario:.2f}")
```

A empresa "Tintas & Curvas S.A." foi contratada para pintar a fachada lateral de um novo centro cultural. O design da parede segue uma linha moderna, onde o topo da parede é definido pela curva y = sin(x) + 2 em um intervalo de 10 metros. O gerente de compras precisa saber a área exata dessa parede para comprar a quantidade certa de tinta. Se ele comprar a menos, a obra para; se comprar a mais, há desperdício de capital. Para resolver isso tecnicamente, utilizaremos a estratégia de dividir a parede em finas faixas verticais e calcular a área de cada uma delas e realizar a soma. Quanto mais finas essas faixas, mais perto do valor correto, portanto é pedido um programa para a realização desta soma.

Tarefa
Desenvolva um programa que calcule a área total da parede e a quantidade de latas de tinta necessárias. Para tal:

1. Crie uma função chamada altura_parede(x) que retorne o valor de math.sin(x) + 2. (Importe o módulo math)
2. Divida o comprimento de 10 metros em n fatias (retângulos).
- A largura de cada fatia será 10 / n.
- A área de cada fatia é altura_parede(x) * largura.
3. Calcule a quantidade de tinta:
- Considere que 1 litro de tinta pinta 5 m^2.
- Cada lata de tinta tem 3,6 litros.
- O programa deve informar quantas latas (número inteiro) devem ser compradas. (Dica: use math.ceil() para arredondar para cima).

Entrada
O usuário deve inserir o número de fatias (n) para o cálculo. (Testar com 10, 100 e 1000 para ver a precisão aumentar).

Saída
O programa deve exibir:

A área total aproximada da parede.
A quantidade total de tinta necessária (em litros).
O número mínimo de latas de 3,6L a serem compradas.

Curiosidade: Parabéns! Você acabou de realizar uma integral.

```
from math import sin, ceil

def altura_parede(x):
    return sin(x) + 2


n = int(input('Digite o total de fatias: '))
dlargura = 10/n

area_total = 0
for i in range(n):
    x = i * dlargura
    area_total += altura_parede(x) * dlargura

total_tinta = area_total/5
latas_tinta = ceil(total_tinta/3.6)

print("Área total: ", area_total)
print("Quantidade minima de litros de tinta: ", total_tinta)
print("Número minimo de latas necessárias: ", latas_tinta)
```

## Exercício 10

A equipe de Produtos Financeiros do Banco InvesteMais está lançando um novo tipo de aplicação de renda fixa, visando atrair investidores que buscam retornos de longo prazo. Eles necessitam de um programa em Python, a ser executado em blocos no Deepnote, que simule o crescimento do capital ao longo do tempo. Este programa deve usar o capital inicial informado pelo cliente, a taxa mensal de juros e o período de investimento, gerando o valor exato a ser pago no resgate de acordo com as regras estabelecidas.

Tarefa
Desenvolva um programa que calcule o montante final de um investimento financeiro considerando juros compostos mensais, aplicando regras condicionais de bônus ou penalidade sobre o capital. Para melhorar a qualidade deste código, é exigido que ele seja realizado majoritariamente dentro de funções.

Entrada
O programa receberá três valores numéricos: o capital principal inicial, a taxa de juros mensal (em formato decimal) e a duração do investimento em meses. O capital principal será um número de ponto flutuante positivo, a taxa mensal será um número de ponto flutuante positivo, e a duração em meses será um número inteiro positivo.

Saída
O programa deve exibir um único valor numérico, representando o montante final do investimento. O valor final deve ser truncado para apresentar apenas duas casas decimais, ignorando quaisquer dígitos adicionais. As seguintes regras devem ser aplicadas sequencialmente para o cálculo:

1. Regra de Penalidade: Se a duração total do investimento for estritamente inferior a 3 meses, o capital não renderá juros. Em vez disso, uma penalidade fixa de 2% do capital principal inicial será deduzida. O valor final será o capital principal inicial menos essa penalidade.
2. Regra de Bônus: Se a duração total do investimento for igual ou superior a 18 meses, um bônus adicional equivalente a 0.5% do capital principal inicial é somado ao capital principal antes da aplicação dos juros compostos. Esta regra é prioritária e não afeta a aplicação da Regra de Penalidade, pois as condições de duração são mutuamente exclusivas.
3. Para durações que não se enquadram nas regras 2 ou 3 (ou seja, 3 meses <= duração < 18 meses), ou após a aplicação do bônus (se aplicável), o montante final é calculado aplicando a taxa mensal ao capital a cada mês sucessivo, pelo número total de meses de investimento.

Regra de Juros Compostos
A fórmula dos juros compostos é:

M = C * (1 + i)^t

Onde:

M = Montante (valor final)
C = Capital inicial (valor investido)
i = Taxa de juros (em decimal)
t = Tempo (número de períodos)

Observação:

A taxa i deve estar em formato decimal e ter como base o mesmo período do tempo. Exemplo:

5% = 0,05
10% = 0,10

```
def recolher_entradas_usuario():
    capital_inicial = float(input("Digite o capital inicial: "))
    taxa_juros = float(input("Digite a taxa de juros: "))
    duracao = int(input("Digite a quantidade de meses: "))
    return capital_inicial, taxa_juros, duracao

def calcular_montante(capital_inicial, taxa_juros, duracao):
    return round(capital_inicial * (1+taxa_juros) ** duracao, 2)

def decisao_sobre_tempo(capital_inicial, taxa_juros, duracao):
    if duracao < 3:
        return round(capital_inicial * 0.98, 2)
    elif duracao < 18:
        return calcular_montante(capital_inicial, taxa_juros, duracao)
    else:
        capital_inicial = capital_inicial*1.05
        return calcular_montante(capital_inicial, taxa_juros, duracao)

def resultado():
    capital_inicial, taxa_juros, duracao = recolher_entradas_usuario()
    montante = decisao_sobre_tempo(capital_inicial, taxa_juros, duracao)
    print(f"O resultado da operação é de: R${montante:.2f}")

resultado()
```

Auditoria de Estoques em Centros de Distribuição A empresa de logística 'GlobalLog' opera dois armazéns principais na região metropolitana. Ele possui duas listas de códigos de identificação e precisa de uma ferramenta automatizada para encontrar os itens em comum.

Tarefa
Escreva uma função que, recebe duas listas do usuário e retorna uma nova chamada lista_intersecao contendo apenas os elementos que estão presentes em ambas as listas. Para melhorar a replicabilidade do código, é necessário que a análise das interseções seja realizada dentro de uma função.

Entrada
A função receberá duas listas. A primeira representa os códigos dos produtos no Armazém A e a segunda os códigos no Armazém B.

Saída
O programa deve exibir a lista lista_intersecao contendo os elementos comuns.

```
def receber_lista_produtos():
    lista = list(map(int, input("Digite os códigos do Armazém A separados por espaço: ").split())) 
    return lista

def intersecao_produtos(produtos_a, produtos_b):
    produtos_comum = []
    for prod in produtos_a:
        if prod in produtos_b:
            produtos_comum.append(prod)
    return produtos_comum

def executar_programa():
    lista_produtos_a = receber_lista_produtos()
    lista_produtos_b = receber_lista_produtos()
    produtos_comum = intersecao_produtos(lista_produtos_a, lista_produtos_b)
    print("Os produtos comuns são: ", produtos_comum)

executar_programa()
```

## Exercício 12

O departamento de computação científica da 'QuantumLab' precisa validar modelos de simulação probabilística. Uma das técnicas mais famosas é o Método de Monte Carlo, que utiliza a geração de pontos aleatórios para estimar valores matemáticos. O desafio proposto pela equipe é calcular o valor de pi simulando o lançamento de dardos em um alvo quadrado que contém um círculo inscrito.

Tarefa
Desenvolva um programa que estime o valor de pi utilizando o método de Monte Carlo.

A Lógica do Experimento
- Imagine um círculo de raio r = 1 inscrito em um quadrado de lado 2.
- A área do círculo é pi × r² = pi.
- A área do quadrado é 2² = 4.
- A razão entre as áreas é pi / 4.
- Se gerarmos muitos pontos aleatórios (x, y) dentro do quadrado, a proporção de pontos que caem dentro do círculo em relação ao total de pontos será aproximadamente pi / 4. Portanto:
- pi ≈ 4 × (pontos dentro do círculo / total de pontos)

Obs.: Para gerar os pontos aleatórios, utilize a função random.random() da biblioteca random. Essa função gera um número decimal entre 0 e 1.

Entrada
Um número inteiro positivo representando o número total de pontos (tentativas) a serem sorteados. (Dica: Use valores altos como 100000 para uma melhor precisão.)

Saída
O programa deve exibir o valor estimado de Pi para 10, 100, 1000, 10000 e 100000 pontos.

```
import random
import math

def ponto_dentro_circulo(x,y):
    hip = math.sqrt(x ** 2 + y ** 2)
    return 1 if hip < 1 else 0

def monte_carlo(qnt):
    pontos_no_circulo = 0
    for i in range(0, qnt):
        x = random.random()
        y = random.random()
        pontos_no_circulo += ponto_dentro_circulo(x,y)
    pi = 4 * (pontos_no_circulo/qnt)
    return pi

repetir = [10, 100, 1000, 10000, 100000]
for rep in repetir:
    print(f"O valor estimado de pi para {rep} pontos: {monte_carlo(rep)}")
```

## Exercício 13

Sistema de Avaliação 'EducaDigital'
O portal de educação 'EducaDigital' realiza simulados nacionais para estudantes de ensino médio. Após a realização das provas, o sistema precisa comparar o desempenho entre candidatos que concorrem às mesmas vagas. Para fins de relatório pedagógico, o sistema deve analisar dois cartões de respostas e determinar qual dos dois alunos obteve a maior pontuação, comparando item a item com um gabarito oficial ou comparando-os entre si.

Tarefa
Crie um programa que compare duas listas de strings, onde cada lista representa as respostas de um aluno (ex: 'A', 'B', 'C'). O programa deve verificar, posição por posição, em quais questões os alunos deram a mesma resposta e, ao final, informar quem teve o maior número de acertos baseando-se em um gabarito fornecido.

Entrada
Três listas de caracteres: a primeira é o gabarito_oficial, a segunda é respostas_aluno_1 e a terceira é respostas_aluno_2. Todas as listas possuem o mesmo tamanho.

Saída
O programa deve exibir o número de acertos de cada aluno e uma mensagem indicando qual aluno venceu (ex: "Aluno 1 venceu", "Aluno 2 venceu" ou "Empate").

```
def receber_lista(tipo):
    lista = list(input(f"Digite {tipo}: ").split())
    return lista

gabarito_oficial = receber_lista("o Gabarito Oficial")
respostas_aluno_1 = receber_lista(" as respostas do aluno 1")
respostas_aluno_2 = receber_lista(" as respostas do aluno 2")

def contar_respostas_certas(gabarito, resposta):
    acertos = 0
    for rep in range(0, len(gabarito)):
        acertos += 1 if gabarito[rep] == resposta[rep] else 0
    return acertos

acertos_aluno_1 = contar_respostas_certas(gabarito_oficial, respostas_aluno_1)
acertos_aluno_2 = contar_respostas_certas(gabarito_oficial, respostas_aluno_2)

print("Número de acertos do aluno 1:", acertos_aluno_1)
print("Número de acertos do aluno 2:", acertos_aluno_2)

if acertos_aluno_1 > acertos_aluno_2:
    print("Aluno 1 venceu")
elif acertos_aluno_1 == acertos_aluno_2:
    print("Empate")
else:
    print("Aluno 2 venceu")
```

## Exercício 14

Segurança e Integridade de Dados CadastraisA plataforma de serviços governamentais 'GovFácil' está atualizando sua base de dados de usuários. Para evitar fraudes e erros de digitação, é obrigatório que todo Cadastro de Pessoa Física (CPF) inserido passe por uma validação rigorosa seguindo o algoritmo oficial da Receita Federal (cálculo dos dígitos verificadores). O analista de sistemas deve garantir que apenas CPFs matematicamente válidos sejam aceitos no sistema de registro.

Tarefa
Desenvolva uma função que receba uma string contendo um CPF no formato XXX.XXX.XXX-XX e verifique se o CPF é válido.

Entrada
Uma string composta por 11 caracteres numéricos.

Saída
O programa deve exibir a mensagem "CPF Válido" se os dígitos verificadores estiverem corretos, ou "CPF Inválido" caso contrário.

Regras Resumidas para Validação de CPF:
1. O CPF deve conter exatamente 11 dígitos numéricos.
2. Não pode ser uma sequência de números iguais (ex: 11111111111, 22222222222).
3. Os 9 primeiros dígitos formam a base do CPF.
4. O 10º dígito é o primeiro dígito verificador, calculado com pesos de 10 a 2.
5. O 11º dígito é o segundo dígito verificador, calculado com pesos de 11 a 2.
6. Para cada dígito verificador:
- Multiplicar os números anteriores pelos pesos.
- Somar os resultados.
- Calcular o resto da divisão por 11.
- O dígito verificador é 11 - o resto calculado acima.
- Se o resultado for 10 ou 11, considerar 0. -Atenção: para o primeiro dígito verificador são utilizados os 9 - dígitos base, para o segundo dígito verificador são utilizados os 9 dígitos base E o primeiro dígito verificador
7. O CPF é válido se os dois dígitos calculados forem iguais aos informados.

```
cpf = input("Digite o CPF:")
def validador_cpf(cpf):
    if not cpf.isdigit():
        return False

    if len(cpf) != 11:
        return False

    if cpf == cpf[0]*11:
        return False
    
    base = cpf[:9]
    def gerar_digito_verificador(base):
        soma = 0
        for i in range(0, len(base)):
            soma += (int(base[i]) * (len(base)+1-i))
        digito_vericador = soma%11
        if digito_vericador > 9:
            return 0
        else:
            return digito_vericador
    base = cpf[:9]
    digito_verificador_1 = gerar_digito_verificador(base)
    base = base + str(digito_verificador_1)
    digito_verificador_2 = 11 - gerar_digito_verificador(base)
    
    return True if cpf[-2:] == str(digito_verificador_1) + str(digito_verificador_2) else False

if validador_cpf(cpf):
    print("CPF Valido")
else:
    print("CPF Inválido")
```

## Exercício 15

Uma organização comunitária local, 'Esperança Solidária', está realizando uma campanha para arrecadar fundos e reformar o parquinho de uma praça pública. Os voluntários da campanha registram cada doação recebida individualmente, com o objetivo de atingir uma meta de arrecadação predefinida. Para otimizar o acompanhamento, eles precisam de uma ferramenta que calcule o total arrecadado. O programa em Python, desenvolvido em blocos no Deepnote, permitirá que os voluntários saibam em tempo real se a meta de 5000 reais foi alcançada ou superada, garantindo transparência e eficiência na gestão dos recursos doados pela comunidade.

Tarefa
Elabore um programa que gerencie uma campanha de arrecadação, somando doações individuais até que uma meta predefinida seja alcançada ou superada. O programa deve monitorar o progresso e apresentar o resultado final com base no cenário alcançado.

Entrada
A entrada consiste em uma sequência de informações. A primeira informação é um número inteiro que representa o valor da meta de arrecadação a ser alcançada. As informações subsequentes são uma série de números reais, cada um correspondendo a uma doação individual recebida. Estes valores serão fornecidos um por um, e o programa deve continuar processando as doações até que a soma total acumulada seja igual ou superior à meta estabelecida. Todos os valores de doação serão positivos e devem ser considerados válidos.

Saída
O programa deve produzir uma única linha de texto. Esta linha deve apresentar o valor total arrecadado, formatado com duas casas decimais. Após o valor, deve haver uma mensagem específica. Se o total arrecadado for exatamente igual à meta, a mensagem deve indicar que a meta foi atingida. Se o total arrecadado for superior à meta, a mensagem deve informar que a meta foi superada, detalhando também o valor exato pelo qual a meta foi excedida, formatado com duas casas decimais.

```
meta = float(input("Digite o valor da meta à ser alcançada"))
arrecadado = 0
while meta >= arrecadado:
    doacao = float(input("Digite o valor da doação obtida: "))
    arrecadado += doacao

print(f"Valor arrecadado: R${arrecadado:.2f}. ", end='')
if meta == arrecadado:
    print("A meta foi atingida")
elif meta < arrecadado:
    print(f"O valor excedido foi de {arrecadado-meta:.2f}" )
```

## Exercício 16

A startup de jogos 'RetroPixels' está desenvolvendo um protótipo de aventura em modo texto. Você foi designado para programar o motor de movimentação do herói. O desafio é levar o personagem através de um quadrante de exploração, garantindo que ele não ultrapasse as fronteiras de segurança e chegue ao objetivo final para completar a missão.

Tarefa
Desenvolva um programa que gerencie a posição de um personagem em um grid de 20 x 20 (coordenadas de 0 a 19). O programa deve rodar em um loop contínuo, aceitando comandos de direção até que o objetivo seja alcançado.

Requisitos Técnicos:

1. Configuração de Variáveis:

O personagem começa na posição: X = 3 e Y = 2.
A bandeira (objetivo) está na posição: X = 15 e Y = 18.

2. Movimentação e Comandos:

O programa deve ler a entrada do usuário:
w: Sobe (diminui Y)
s: Desce (aumenta Y)
a: Esquerda (diminui X)
d: Direita (aumenta X)
Garanta que o comando funcione com letras maiúsculas ou minúsculas.

3. Processamento por Funções:

Crie uma função que receba a posição atual e o comando, retornando a nova posição após validar se o movimento é possível.

4. Renderização do Grid:

Antes do primeiro comando, e após cada comando, o programa deve imprimir o grid de 20 linhas por 20 colunas.
Use os seguintes caracteres:
. (ponto) para espaços vazios.
P (letra P) para a posição atual do Personagem.
B (letra B) para a posição da Bandeira.

5. Regras de Fronteira:

O grid possui limites entre 0 e 19. O personagem não pode sair desses limites. Se o usuário tentar ir para X=-1 ou Y=20, por exemplo, o programa deve exibir uma mensagem de "Colisão com a fronteira!" e manter o personagem no mesmo lugar.

6. Ciclo do Jogo:

O jogo deve continuar até que o personagem chegue na mesma posição da bandeira.

7. Saída de Dados:

Ao chegar no objetivo, exiba: "Objetivo alcançado! Missão cumprida.". Caso o input do usuário seja "sair", o jogo é encerrado.

```
from IPython.display import clear_output

pos_personagem = [3,2]
pos_bandeira = [15,18]

def printar_mapa(pos_personagem, pos_bandeira):
    for y in range(20):
        for x in range(20):
            if [x, y] == pos_personagem:
                print(f'{ "P":2}', end='')
            elif [x, y] == pos_bandeira:
                print(f'{ "B":2}', end='')
            else:
                print(f'{ ".":2}', end='') 
        print('\n', end='')

def movimentacao():
    def validar_digitacao_movimentacao():
        comando = 0
        while comando not in ['w', 's', 'a', 'd']:
            comando = input("Digite um comando: ")
            if comando.lower() == 'sair':
                return comando
            if comando not in ['w', 's', 'a', 'd']:
                print('Movimentação Invalida tente novamente')
        return comando.lower()
    
    def checa_fronteira(comando):
        match comando:
            case 'w': True if pos_personagem[1] == 0 else False
            case 's': True if pos_personagem[1] == 19  else False
            case 'a': True if pos_personagem[0] == 0  else False
            case 'd': True if pos_personagem[0] == 19  else False
        
    def andar(comando):
        match comando:
            case 'w': pos_personagem[1] = pos_personagem[1] - 1
            case 's': pos_personagem[1] = pos_personagem[1] + 1
            case 'a': pos_personagem[0] = pos_personagem[0] - 1
            case 'd': pos_personagem[0] = pos_personagem[0] + 1

    while True:
        comando = validar_digitacao_movimentacao()
        if comando == 'sair':
            return comando

        if checa_fronteira(comando):
            print("Personagem está na fronteira, ele não pode andar para este caminho")
            continue
        else:
            andar(comando)
            break


while True:
    printar_mapa(pos_personagem, pos_bandeira)

    comando = movimentacao()
    if comando == 'sair':
        print("Jogo encerrado pelo usuário")
        break

    if pos_personagem[0] == pos_bandeira[0] and pos_personagem[1] == pos_bandeira[1]:
        print("Objetivo alcançado! Missão cumprida.")
        break
    clear_output()
```
