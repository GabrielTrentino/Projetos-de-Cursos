Desenvolva um programa que receba a descrição textual de uma despesa e seu respectivo valor, classificando-a em uma de três categorias predefinidas com base em termos específicos contidos na descrição. A comparação deve ser insensível a maiúsculas e minúsculas. Caso a descrição contenha palavras de categorias diferentes, siga a ordem de prioridade descrita abaixo.

#### Entrada
A entrada é composta por duas linhas distintas:

1. A primeira linha contém um texto que representa a descrição da despesa. A descrição pode conter letras maiúsculas ou minúsculas.
2. A segunda linha contém um número real que representa o valor da despesa.

#### Saída
O programa deve produzir uma única linha de saída contendo o nome da categoria à qual a despesa pertence, seguido de um hífen e do seu valor. Utilize a seguinte prioridade e lista de termos (todos em letras minúsculas para a verificação):

## Exercício 7

A loja de eletrônicos "TecnoMais" busca otimizar suas vendas implementando um novo sistema de descontos progressivos. Para garantir que os clientes recebam o preço mais justo e a loja mantenha sua competitividade, a gerência definiu diferentes faixas de abatimento com base no valor total da compra. Além disso, clientes que pagam com o **Cartão TecnoMais** têm direito a um abatimento adicional. Um desenvolvedor iniciante deve construir um programa em Python, a ser executado em blocos no ambiente Deepnote, que calcule o valor final a ser pago por um cliente, aplicando automaticamente o desconto correto e o abatimento extra. Este programa permitirá que a "TecnoMais" agilize o processo de vendas e aprimore a satisfação dos seus compradores.

### Tarefa

Desenvolva um algoritmo que determine o valor final de uma compra após a aplicação de um desconto progressivo e de um desconto adicional para pagamentos com o Cartão TecnoMais. O cálculo do desconto é condicional ao valor total da aquisição e ao método de pagamento.

#### Entrada
A entrada é composta por duas linhas:

1. Um único número real representando o valor total bruto da compra. Este valor é sempre positivo e pode conter até duas casas decimais.
2. Uma string que indica se o pagamento é realizado com o Cartão TecnoMais. Use `S` para sim ou `N` para não (ignorar diferenças de maiúsculas/minúsculas).

#### Saída
A saída deve ser um único número real, correspondendo ao valor final da compra após a aplicação dos descontos. O resultado deve ser apresentado com exatamente duas casas decimais, independentemente da precisão original do cálculo. As regras de desconto são as seguintes:

- Compras com valor até R$ 100,00 não recebem desconto.
- Compras entre R$ 100,01 e R$ 500,00 recebem 5% de desconto.
- Compras entre R$ 500,01 e R$ 1000,00 recebem 10% de desconto.
- Compras acima de R$ 1000,00 recebem 15% de desconto.

Clientes que utilizam o Cartão TecnoMais recebem um desconto adicional de 2 pontos percentuais na mesma faixa em que se enquadram, caso já estejam recebendo algum desconto. Por exemplo, se a compra é de R$ 350,00 e o cliente paga com o Cartão TecnoMais, ele recebe 5% + 2% de desconto; se a compra é de R$ 1 200,00 e o cliente paga com o Cartão TecnoMais, ele recebe 15% + 2% de desconto.


```python
valor_bruto = float(input("Digite o valor total bruto da compra: "))
flag_cartao = input("Digite S para sim ou N para não se o pagamento foi realizado com o Cartão TecnoMais: ")

desconto = 2 if flag_cartao.lower() == 's' else 0
if valor_bruto > 1000:
    desconto += 15
elif valor_bruto > 500:
    desconto += 10
elif valor_bruto > 100:
    desconto += 5

valor_final = valor_bruto * (100-desconto)/100
print(f"O valor final da compra é: {valor_final:.2f}")
```

## Exercício 9

Em um jogo de aventura chamado Caverna dos Caminhos Incertos, o jogador explora uma caverna antiga cheia de decisões estranhas e perigos inesperados. A cada etapa, o caminho seguido depende das escolhas anteriores, e nem todas as informações são sempre solicitadas: algumas decisões só fazem sentido se o jogador já tiver seguido por determinado caminho. Seu objetivo é criar um programa em Python, executado em blocos no Deepnote, que simule esse sistema de decisões encadeadas do jogo e determine o resultado final da exploração.

### Tarefa

Desenvolva um programa que conduza a lógica do jogo a partir de uma decisão inicial. Dependendo dessa decisão, o programa deve solicitar novas entradas e, ao final, indicar o destino do jogador na caverna. 

### Entrada

A entrada ocorre em etapas, seguindo a lógica abaixo:

1. A primeira linha contém uma string indicando o caminho inicial escolhido, que pode ser:
   - `Esquerda`
   - `Direita`

2. Se o caminho inicial for `Esquerda`, o programa deve ler uma segunda entrada:
   - Um número inteiro representando o nível de coragem do jogador (0 a 100).

3. Se o caminho inicial for `Direita`, o programa deve ler uma segunda entrada:
   - Uma string indicando se o jogador possui uma tocha, podendo ser `Sim` ou `Nao`.

4. Dependendo das escolhas anteriores, uma terceira entrada pode ser necessária:
   - Se o caminho for `Esquerda` e o nível de coragem for menor que 80, ler um número inteiro indicando o nível de vida restante (0 a 100).
   - Se o caminho for `Direita` e o jogador possuir tocha (`Sim`), ler um número inteiro indicando o tempo restante da tocha em minutos.

Assuma que todas as entradas estarão dentro dos valores esperados.

### Saída

O programa deve imprimir uma única linha de texto indicando o **resultado final da exploração**, conforme as regras abaixo:

- Se o caminho inicial for `Esquerda`:
  - Se a coragem for maior ou igual a 80, o resultado é `Tesouro Encontrado`.
  - Caso contrário:
    - Se a vida restante for maior ou igual a 30, o resultado é `Fuga Bem-Sucedida`.
    - Se a vida restante for menor que 30, o resultado é `Derrota na Caverna`.

- Se o caminho inicial for `Direita`:
  - Se o jogador **não** possui tocha (`Nao`), o resultado é `Perdido na Escuridao`.
  - Caso contrário:
    - Se o tempo da tocha for maior que 10 minutos, o resultado é `Caminho Seguro`.
    - Se o tempo da tocha for 10 minutos ou menos, o resultado é `Avanco Arriscado`.

```python
caminho = input("Digite o caminho à ser seguido: Direita ou Esquerda: ")

if caminho.lower() == "esquerda":
    coragem = int(input("Digite o nível de coragem: "))
    if coragem < 80:
        vida_restante = int(input("Digite a quantidade de vida restante: "))
        if vida_restante >= 30:
            print("Fuga Bem-Sucedida")
        else: 
            print("Derrota na Caverna")
    else:
        print("Tesouro Encontrado.")

elif caminho.lower() == "direita":
    possui_tocha = input("O jogador possui uma tocha? ")
    if possui_tocha.lower() == "sim":
        tempo_restante = int(input("Digite o tempo restante da tocha: "))
        if tempo_restante >= 10:
            print("Caminho Seguro")
        else:
            print("Avanco Arriscado")
    else:
        print("Perdido na Escuridao")
```

## Exercício 10

No jogo de ação BattleQuest o desempenho do jogador em uma partida é medido por diferentes fatores do gameplay. Ao final da partida, o sistema precisa calcular uma pontuação final com base em: inimigos derrotados, objetivos conquistados, mortes sofridas e tempo de partida. A partir dessa pontuação, o jogo decide se o jogador alcança o status especial Jogador Elite. Seu programa em Python, executado em blocos no Deepnote, deve calcular essa pontuação seguindo regras por faixas e, ao final, informar se o jogador atingiu o status.

### Tarefa

Desenvolva um programa que:

1. Calcule a pontuação final do jogador somando quatro componentes de pontuação (um para cada fator).
2. Verifique se a pontuação final é maior que 100 para definir se o jogador ganha o status "Jogador Elite".

#### Entrada

Serão fornecidos quatro valores em linhas separadas, na seguinte ordem:

- Um inteiro representando o número de inimigos derrotados.
- Um inteiro representando o número de objetivos conquistados.
- Um inteiro representando o número de mortes sofridas (quanto maior, pior).
- Um inteiro representando o tempo de partida em minutos.

Considere que todos os valores são inteiros e não negativos.

#### Regras

A pontuação final é calculada assim:

1. Pontos por inimigos derrotados
    - Se inimigos < 10: **+10 pontos**
    - Se 10 ≤ inimigos ≤ 29: **+30 pontos**
    - Se inimigos ≥ 30: **+50 pontos**

2. Pontos por objetivos conquistados
    - Se objetivos == 0: **+0 pontos**
    - Se 1 ≤ objetivos ≤ 2: **+25 pontos**
    - Se objetivos ≥ 3: **+60 pontos**

3. Penalidade por mortes sofridas
    - Se mortes < 5 : **0 pontos**
    - Se 5 ≤ mortes ≤ 10: **-10 pontos**
    - Se mortes > 10: **-30 pontos**

4. Pontos por tempo de partida (em minutos)
    - Se tempo < 5: **+25 pontos**
    - Se 5 ≤ tempo ≤ 14: **+15 pontos**
    - Se tempo ≥ 15: **+5 pontos**

A pontuação final é a soma:
`pontos_inimigos + pontos_objetivos + penalidade_mortes + pontos_tempo`

Caso obtenha mais do que 100 pontos, o jogador é classificado como `Jogador Elite`

#### Saída
O programa deve imprimir uma linha com a pontuação final do jogador, e, caso o jogador seja "Jogador Elite" uma segunda linha indicando, e parabenizando-o


```python
inimigos = int(input("Digite a quantidade de inimigos derrotados: "))
objetivos = int(input("Digite a quantidade de objetivos concluidos: "))
mortes = int(input("Digite a quantidade de mortes sofridas: "))
tempo = int(input("Digite a quantidade de tempo da partida: "))

pontos = 0
if inimigos >=30:
    pontos += 50
elif inimigos >= 10:
    pontos += 30
else:
    pontos += 10

if objetivos >=3:
    pontos += 60
elif objetivos >= 1:
    pontos += 25

if mortes >=10:
    pontos -= 30
elif mortes >= 5:
    pontos -=10

if tempo >=15:
    pontos += 5
elif tempo >= 5:
    pontos += 15
else:
    pontos += 25

print("Sua pontuação final é: ", pontos)
if pontos > 100:
    print("Você foi classificado como Jogador Elite pela pontuação, parabéns.")
```

### Tarefa

Desenvolva um programa que determine o nível de alerta meteorológico de uma região com base nos dados climáticos informados. As condições devem ser verificadas em ordem hierárquica, garantindo que alertas mais graves tenham prioridade sobre os demais.

A entrada consiste em quatro informações fornecidas em linhas sucessivas, nesta ordem:

1. Um número inteiro representando a velocidade do vento em km/h.
2. Um número inteiro representando a temperatura em graus Celsius.
3. Um número inteiro representando a umidade do ar em percentual (0 a 100).
4. Uma string representando o tipo de precipitação, que pode ser exatamente: `Chuva`, `Neve`, `Granizo` ou `Nenhum`

#### Saída

O programa deve produzir uma única linha de texto, contendo exatamente uma das seguintes mensagens:

- `ALERTA VERMELHO (Perigo Iminente)`
- `ALERTA LARANJA (Cuidado)`
- `ALERTA AMARELO (Atenção)`
- `ALERTA AZUL (Observação)`
- `SEM ALERTA`

As regras abaixo devem ser avaliadas na ordem apresentada:

1. ALERTA VERMELHO (Perigo Iminente)
Emitido se qualquer das condições abaixo for verdadeira:
    - Velocidade do vento maior que 80 km/h
    - Precipitação igual a `Granizo` e velocidade do vento maior que 60 km/h

2. ALERTA LARANJA (Cuidado)
Emitido se não for ALERTA VERMELHO e qualquer das condições abaixo ocorrer:
    - Temperatura menor que -10°C
    - Precipitação igual a `Neve` e velocidade do vento maior que 40 km/h
    - Precipitação igual a `Neve` e temperatura menor que 0°C
    - Velocidade do vento maior que 50 km/h

3. ALERTA AMARELO (Atenção)
Emitido se não for nenhum alerta anterior e qualquer das condições abaixo ocorrer:
    - Precipitação igual a `Chuva` e velocidade do vento maior que 40 km/h e umidade maior que 90%
    - Velocidade do vento maior que 20 km/h
    - Umidade maior que 80%

4. ALERTA AZUL (Observação)
Emitido se não for nenhum alerta anterior e qualquer das condições abaixo ocorrer:
    - Precipitação igual a `Nenhum` e temperatura menor que 5°C e velocidade do vento maior que 30 km/h
    - Temperatura menor que 10°C
    - Umidade maior que 75%
    - Precipitação igual a `Chuva`

5. SEM ALERTA
Emitido se nenhuma das condições anteriores for satisfeita.

O programa será executado em blocos no Deepnote e deverá interpretar corretamente os dados fornecidos para produzir o resultado esperado.

### Tarefa

Desenvolva um programa que funcione como uma calculadora básica. O programa deve receber dois números e uma indicação da operação matemática desejada e, a partir disso, calcular e exibir o resultado correspondente.

### Entrada

A entrada do programa consiste em três linhas:

1. Um número real representando o primeiro valor da operação.
2. Um número real representando o segundo valor da operação.
3. Uma string indicando a operação matemática a ser realizada, que pode ser exatamente uma das seguintes:
   - `soma`
   - `subtracao`
   - `multiplicacao`
   - `divisao`

Assuma que os valores numéricos são válidos e que, no caso da operação de divisão, o segundo número será diferente de zero.

### Saída

O programa deve produzir uma única linha contendo o resultado da operação, apresentado como um número real.

```python
num1 = float(input("Digite o primeiro numero: "))
num2 = float(input("Digite o segundo numero: "))
operacao = input("Digite a operacao desejada: ")

match operacao.lower():
    case 'soma':
        resultado = num1 + num2
    case 'subtracao':
        resultado = num1 - num2
    case 'multiplicacao':
        resultado = num1 * num2
    case 'divisao':
        resultado = num1/num2

print("O resultado da operação é: ", resultado)
```

## Exercício 14

Em um hospital, a equipe de triagem precisa verificar rapidamente se um paciente pode receber doação de sangue de outra pessoa. Para isso, o sistema deve avaliar o tipo sanguíneo (A, B, AB, O) e o fator Rh (+ ou -) do receptor e do doador e indicar se a doação é compatível.

### Tarefa

Desenvolva um programa que receba os dados de duas pessoas e determine se a Pessoa 2 pode doar sangue para a Pessoa 1, seguindo as regras de compatibilidade indicadas.

#### Entrada

A entrada consiste em quatro linhas, na seguinte ordem:

    1. Uma string com o tipo sanguíneo da Pessoa 1 (receptor): `A`, `B`, `AB` ou `O`.
    2. Uma string com o fator Rh da Pessoa 1: `+` ou `-`.
    3. Uma string com o tipo sanguíneo da Pessoa 2 (doador): `A`, `B`, `AB` ou `O`.
    4. Uma string com o fator Rh da Pessoa 2: `+` ou `-`.

Assuma que as entradas sempre estarão dentro desses valores.

#### Saída

O programa deve imprimir uma única linha:

- `Doacao compativel` se a Pessoa 2 pode doar para a Pessoa 1.
- `Doacao incompatível` caso contrário.

#### Regras de compatibilidade para doação de sangue

A compatibilidade entre doador e receptor deve ser avaliada em duas etapas independentes:  
primeiro pelo tipo sanguíneo (A, B, AB, O) e depois pelo fator Rh (+ ou -).  
A doação só é possível se ambas as compatibilidades forem satisfeitas.

1. Compatibilidade por tipo sanguíneo (ABO)

    - Receptor do tipo A pode receber sangue dos tipos: `A` e `O`
    - Receptor do tipo B pode receber sangue dos tipos: `B` e `O`
    - Receptor do tipo AB pode receber sangue dos tipos: `A`, `B`, `AB` e `O`
    - Receptor do tipo O pode receber sangue apenas do tipo: `O`

2. Compatibilidade por fator Rh

    - Receptor com fator Rh positivo (+) pode receber sangue de doadores com Rh `+` ou `-`
    - Receptor com fator Rh negativo (-) pode receber sangue apenas de doadores com Rh `-`

**Regra final:** A doação de sangue é considerada compatível somente se o tipo sanguíneo do doador for compatível com o tipo do receptor e o fator Rh do doador for compatível com o fator Rh do receptor.

```python
tipo1 = input("Digite o tipo sanguíneo da Pessoa 1 (receptor): ")
rh1 = input("Digite o fator Rh da pessoa 1: ")
tipo2 = input("Digite o tipo sanguíneo da Pessoa 2 (doador): ")
rh2 = input("Digite o fator Rh da pessoa 2: ")

flag_abo = False
if   tipo1.lower() == 'a'  and tipo2.lower() in ['a', 'o']:
    flag_abo = True
elif tipo1.lower() == 'b'  and tipo2.lower() in ['b', 'o']:
    flag_abo = True
elif tipo1.lower() == 'ab':
    flag_abo = True
elif tipo1.lower() == 'o'  and tipo2.lower() == 'o':
    flag_abo = True

flag_rh = False
if rh1 == '-' and rh2 == '-':
    flag_rh = True
elif rh1 == '+':
    flag_rh = True

if flag_rh + flag_abo == 2:
    print("Doacao compativel")
else:
    print("Doacao incompativel")
```

## Exercício 16

Uma empresa está desenvolvendo um sistema interno de **triagem cadastral (KYC)** para decidir se um novo cliente pode ser aprovado automaticamente ou se precisará passar por revisão manual. O processo varia de acordo com o **tipo de cliente** (Pessoa Física ou Pessoa Jurídica) e, dependendo das escolhas iniciais, **novas informações devem ser solicitadas**.

Seu programa em Python, executado em blocos no Deepnote, deve seguir exatamente o fluxo de decisões descrito abaixo para determinar o resultado final do cadastro.

### Tarefa

Desenvolva um programa que avalie um cadastro de cliente por etapas.  
Dependendo do tipo de cliente informado inicialmente, o programa deve solicitar informações adicionais específicas e, ao final, indicar o status do cadastro.

O programa deve utilizar **condicionais aninhadas**, pois nem todas as entradas são solicitadas em todos os casos.

### Entrada

A entrada ocorre em etapas, na seguinte lógica:

1. A primeira linha contém uma string indicando o **tipo de cliente**, que pode ser:
   - `PF` (Pessoa Física)
   - `PJ` (Pessoa Jurídica)

2. Se o tipo for `PF`, o programa deve ler:
   - Uma string representando o **CPF**, contendo apenas números.
   - Um número inteiro representando a **idade** do cliente.

3. Se o tipo for `PJ`, o programa deve ler:
   - Uma string representando o **CNPJ**, contendo apenas números.

4. Dependendo das validações anteriores, uma nova entrada pode ser necessária:
   - Se o tipo for `PF` e o CPF **não tiver exatamente 11 caracteres**, ler uma string indicando se há **documentação complementar**, `S` ou `N`.
   - Se o tipo for `PJ` e o CNPJ **não tiver exatamente 14 caracteres**, ler uma string indicando se a empresa possui **registro provisório**, `S` ou `N`.

Caso a primeira linha não seja válida, exiba uma mensagem de erro informando ao usuário. Assuma que as demais entradas seguem o formato esperado (strings ou números inteiros).

### Saída

O programa deve imprimir uma única linha com o **status final do cadastro**, conforme as regras abaixo:

- Se o tipo for `PF`:
  - Se o CPF tiver exatamente 11 caracteres **e** a idade for maior ou igual a 18, imprimir `Cadastro Aprovado`.
  - Caso contrário:
    - Se houver documentação complementar (`S`), imprimir `Cadastro em Revisao` **e** a idade for maior ou igual a 18.
    - Caso contrário, imprimir `Cadastro Recusado`.

- Se o tipo for `PJ`:
  - Se o CNPJ tiver exatamente 14 caracteres, imprimir `Cadastro Aprovado`.
  - Caso contrário:
    - Se houver registro provisório (`S`), imprimir `Cadastro em Revisao`.
    - Caso contrário, imprimir `Cadastro Recusado`.

```python
tipo_cliente = input("Digite o tipo de cliente: ")

if tipo_cliente.upper() in ["PF", "PJ"]:

    if tipo_cliente.upper() == "PF":
        cgcent = input("Digite o CPF da pessoa: ")
        idade = int(input("Digite a idade: "))

        if len(cgcent) != 11: 
            doc_complementar = input("Há documentação complementar? ")
            if doc_complementar.upper() == "S" and idade >= 18:
                print("Cadastro em Revisão")
            else:
                print("Cadastro Recusado")
        else:
            if idade >= 18:
                print("Cadastro Aprovado")
            else:
                print("Cadastro Recusado")

    elif tipo_cliente.upper() == "PJ":
        cgcent = input("Digite o CNPJ da empresa: ")

        if len(cgcent) != 14:
            reg_provisorio = input("A empresa possui registro provisório? ")
            if reg_provisorio.upper() == "S":
                print("Cadastro em Revisão")
            else:
                print("Cadastro Recusado")

        else:
            print("Cadastro Aprovado")

else:
    print("Recomeçe o programa, tipo de cliente não especificado PF ou PJ.")
```
